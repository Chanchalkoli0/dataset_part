```python
import pandas as pd 
```

```python
# Load dataset
df = pd.read_csv("sales_data.csv")

# Handle missing values
df["Quantity"] = df["Quantity"].fillna(0)
df["Price"] = df["Price"].fillna(df["Price"].mean())

# Recalculate total sales for accuracy
df["Computed_Total_Sales"] = df["Quantity"] * df["Price"]

# Metrics
total_sales = df["Computed_Total_Sales"].sum()
best_product_quantity = df.groupby("Product")["Quantity"].sum().idxmax()
best_product_revenue = df.groupby("Product")["Computed_Total_Sales"].sum().idxmax()
average_order_value = df["Computed_Total_Sales"].mean()
top_region = df.groupby("Region")["Computed_Total_Sales"].sum().idxmax()

# Output report
print("📊 Sales Analysis Report")
print("-----------------------")
print(f"Total Sales: ₹{total_sales:.2f}")
print(f"Best-Selling Product (Quantity): {best_product_quantity}")
print(f"Best-Selling Product (Revenue): {best_product_revenue}")
print(f"Average Order Value: ₹{average_order_value:.2f}")
print(f"Top Revenue Region: {top_region}")
```
